```python
from llm_rag_ai_agent.config import settings
from openai import OpenAI
from IPython.lib.pretty import pretty
```

```python
open_api_client = OpenAI(
    api_key=settings.nvidia_api_key,
    base_url=settings.nvidia_base_url)
open_api_client
```

```python
from openai.resources.containers.files import content
message = input("Enter your message")
role = "user"
response = open_api_client.chat.completions.create(
    model=settings.nvidia_model,
    messages=[
        dict(
            role=role,
            content=message
        )
    ]
)
```

```python
print(response.choices[0].message.content)
```

```python
characters = {
  "Sherlock Holmes": "Your are a help full and brilliant detective known for his exceptional observation, deductive reasoning, and ability to solve complex mysteries.",
  "Spider-Man": "Your are a help full and  young superhero with spider-like abilities who balances everyday life with the responsibility of protecting people from danger.",
  "Batman": "Your are a help full and  highly intelligent and skilled vigilante who uses detective skills, technology, and strategic thinking to fight crime.",
  "Iron Man": "Your are a help full and  genius inventor and billionaire who uses advanced technology, engineering, and his Iron Man suit to become a superhero.",
  "James Bond": "Your are a help full and  highly trained secret agent known for his intelligence, combat skills, confidence, and ability to handle dangerous missions."
}
```

```python
from IPython.display import display
import ipywidgets as widgets
character_names = list(characters.keys())
dropdown = widgets.Dropdown(
    options=characters,
    description="Character:"
)

display(dropdown)
```

```python
user_query = input("Enter the query ?")
system_prompt = characters.get(character_names[dropdown.index], "")
response = open_api_client.chat.completions.create(
    model=settings.nvidia_model,
    messages=[
        {"role": "user", "content":user_query},
        {"role": "system", "content":system_prompt}
    ]
)
response
```
